## The project in one paragraph
This project analysed 113 months of HMRC tax receipts (April 2017 to August 2026), about £6.95 trillion in total and an average of about £61.5bn a month. I cleaned the data, explored it, tested two hypotheses, built and compared two prediction models, and shared the results in an interactive Power BI dashboard.

## Answers to the business questions

### 1. Which taxes bring in the most money?
Income Tax is the biggest, at about £20.1bn a month, followed by National Insurance (£13.5bn), VAT (£12.4bn) and Corporation Tax (£5.9bn). These four bring in about 85% of everything collected.

### 2. Is there a seasonal pattern, and when does each tax arrive?
Yes, a strong one. January is about 54% above a typical month and May is about 15% below. Each tax has its own rhythm: VAT peaks every quarter (January, April, July, October), Corporation Tax peaks in March, June, September and December, and Capital Gains Tax is almost all collected in January. January stands out because Income Tax, Capital Gains Tax and VAT all peak together.

### 3. How have receipts changed since 2017?
They grew from about £593bn in 2017/18 to about £938bn in 2025/26. 2020/21 is the only year they fell (down 7.8%), followed by a rebound of 22.5% the year after, and growth of between 3.6% and 10.0% every year since. These figures are not adjusted for inflation.

### 4. Do some taxes move together?
Income Tax and National Insurance move together most closely (correlation 0.82). The link is weaker once the shared upward trend is removed (0.38), so it is real but moderate.

### 5. Can we predict future monthly receipts?
Reasonably well. The linear regression model, using a time trend and the month of the year, missed by about £5.4bn a month on average (6.2%) on the 12 months it had never seen. That beats both the simple guess of "same month last year" (about £6.6bn, 8.1%) and the random forest (about £9.7bn, 10.0%, with an R-squared of -0.13).

## Hypothesis results
- **January effect:** the null hypothesis (January receipts are the same as other months) was rejected. January averaged about £95.5bn against about £58.6bn for the other months (p about 0.0001), and the result held after removing the trend.
- **Income Tax and National Insurance:** the null hypothesis (no positive correlation) was rejected, both on monthly receipts (r = 0.82) and on yearly growth rates (r = 0.38).

## Why linear regression won
The random forest cannot predict above the highest value it saw in training, so it flattens when receipts keep rising. Linear regression keeps following the trend, uses only two clues that our own analysis had already proven useful, and is easier to explain to a finance team.

## Recommendations
1. **Plan around the peaks.** Expect a very large inflow every January, VAT peaks in January, April, July and October, and Corporation Tax peaks in March, June, September and December.
2. **Use the model for rough monthly planning only,** and add a safety margin for January. In January 2026 it predicted about £107.5bn against an actual £126.4bn.
3. **Forecast Income Tax and National Insurance together,** but do not rely on one to predict the other, because the link is moderate once the shared trend is removed.
4. **Improve the model next** by adding the previous year's value for the same month as a clue, training only on the period after Covid, and adjusting for inflation.

## Limitations
- The latest months are provisional and may be revised.
- Figures are not adjusted for inflation.
- Only nine Januaries are in the data, and monthly receipts are not fully independent of each other, so the p-values are best read as indicative.
- The model was tested on only 12 months, so its scores are a rough guide.
- Correlation does not show that one tax causes another.

## Final thought
The data told a clear story: HMRC's receipts depend heavily on four taxes, follow a strong yearly rhythm led by January, and have grown steadily since 2017. A simple, explainable model captures much of that pattern, and the dashboard lets anyone explore it for themselves.
